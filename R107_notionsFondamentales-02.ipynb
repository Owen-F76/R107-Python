## 04- Programmation Orientée Objet côté utilisateur  

**L'exemple de la classe str**

Un objet est une **structure de données**, comme les variables, qui peut elle-même contenir d'autres variables et fonctions.  
En fait, en Python, tout est objet.

### 📚 1. Les méthodes de la classe str

#### 📌 Un premier exemple
🔎 Passer une chaîne de caractères en minuscules.

```python
chaine="RESTONS CALMES !"
chaine.lower() # mettre la chaîne en minuscules
```

#### 📌 Comprendre la structure  
  
L'écriture **chaine.lower()** signifie que la fonction **lower** est une fonction de la variable **chaine** tout comme l'écriture **math.sqrt** signifiait que la fonction **sqrt** était une fonction du module **math**. Dans les deux expressions, le point symbolise l'appartenance : la fonction **lower** appartient à la variable **chaine** tout comme la fonction **sqrt** appartient au module **math**.  
La fonction **lower** est propre aux chaînes de caractères et toutes les chaînes peuvent y faire appel.  

**Remarque**  
L'objet issu de la classe str ne porte pas obligatoirement le nom "chaine" comme le montre l'exemple ci-dessous !

```python
# exemple montrant que la fonction lower() appartient à tout objet issu de la classe str

phrase="SOIS ATTENTIF !"
phrase.lower()
```

```python
# exemple montrant que la fonction n'appartient pas aux objets autres qu'issus de str
phrase=3
phrase.lower()
```

```python
type(phrase)
```

❗️Jusqu'ici, nous avons considéré qu'une variable était issue d'un **type** de donnée. Il faut en fait reformuler comme suit : un **objet** est issu d'une **classe**. La **classe** est une forme de type de donnée sauf qu'elle permet de définir des fonctions et variables propres au type. Ainsi, la fonction **lower** a été définie dans la classe **str**. Les fonctions définies dans une classe sont appelées des **méthodes**.

#### 📌 Petit point d'étape

Les **objets** peuvent contenir des variables ou fonctions appelées **méthodes**.  
On appelle une méthode d'un objet grâce à **objet.methode()**.  (_L'objet phrase contient la méthode lower que l'on appelle grâce à  phrase.lower()_)  

Une **classe** est un modèle qui servira à construire un objet. C'est dans la classe que l'on définira les méthodes propres à l'objet.  

**🛠️ Mécanisme permettant d'appeler la méthode _lower_ d'une chaîne**  
* Les développeurs de Python ont créé la classe **str** utilisée pour créer des chaînes de caractères. Dans cette classe, ils ont défini plusieurs méthodes, comme _lower_, qui pourront être utilisées par n'importe quel objet construit sur cette classe.
* Lorqu'on écrit chaine="RESTONS CALMES !", Python reconnaît qu'il doit créer une chaîne de caractères. Il va donc créer un objet d'après la classe (soit encore le modèle) qui a été défini à l'étape précédente.
* On peut alors appeler toutes les méthodes de la classe **str** depuis l'objet chaine que l'on a créé.

```python
phrase="3" # cette fois 3 est une chaîne de caractères, et non un entier
type(phrase)
```

```python
chaine.title() #mettre toute la chaine en minuscules sauf les premières lettres
```

```python
chaine
```

❗️**Remarque très importante** : lorsqu'on appelle une méthode pour modifier la chaîne, l'objet lui-même n'est pas modifié en mémoire. Seul l'affichage est modifié.

```python
essai=chaine.lower()
print(chaine)
print(essai) # on affecte à essai la chaine en minuscules
```

```python
# test avec l'appel d'une autre méthode
essai=chaine.capitalize() # affiche une majuscule en début de phrase, des minuscules partout ailleurs
print(chaine)
print(essai)
```

### 📚 5. Le parcours de liste

```python
# première méthode : boucle while
ma_liste=['Mercure','Vénus','Terre','Mars','Jupiter','Saturne','Uranus','Neptune']
i=0
while i<len(ma_liste):
    print(ma_liste[i])
    i+=1
```

```python
# deuxième méthode : boucle for
for elt in ma_liste:
    print(elt)
```

```python
# troisième méthode : enumerate
ma_liste=['Mercure','Vénus','Terre','Mars','Jupiter','Saturne','Uranus','Neptune']
for i,elt in enumerate(ma_liste):
    print("En position {} on trouve {}.".format(i,elt))
```

```python
for elt in enumerate(ma_liste):
    print(elt)
```

Avec enumerate, on capture l'indice et l'élément dans deux variables distinctes.

```python
for elt in ma_liste:
    print(elt)
```
