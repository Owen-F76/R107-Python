```python
def personnage_le_plus_puissant(personnages):

    max_niveau = 0
    nom_puissant = ""

    for element in personnages:
        nom = element[0]
        stats = element[2]
        niveau = round(sum(stats)/len(stats),1)
        if niveau > max_niveau:
            max_niveau = niveau
            nom_puissant = element[0]
print (f"Le personnage le plus puissant est {nom_puissant} avec un niveau de {max_niveau}")

personnages = [
    ["Aragorn", "Guerrier", [15, 18 ,12, 8]]
    ["Gandalf", "Mage", [10, 8 ,12, 20]]
    ["Legolas", "Voleur", [14, 16 ,20, 10]]
]

personnage_le_plus_puissant(personnages)
```
